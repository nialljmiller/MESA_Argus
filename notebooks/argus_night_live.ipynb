# An Argus night, live

The field movie, drawn live: a real field observed for one night by simulated Argus and processed by hdps (reduction, WCS, reprojection onto its sky tiles, SFFT difference imaging against a deep reference, detection, alert features), with **one star, E, that you edit**.

* **Field**: M3, M67 or M45 (the Pleiades), each one precomputed hdps night.
* **Band**: β or ρ.
* **Star type** and its sliders: magnitude, amplitude, period (a transient: its timescale).
* **Click** the exposure or the difference image to move star E.
* **Airmass +**: observe the same night lower in the sky.
* **Play** runs through the night; the frame slider goes to any exposure.

Everything except E keeps its photometry and alerts from the real hdps run. E is measured on hdps's images with hdps's own detector, alert features and the real/bogus model; the note under the controls lists what is simplified for it.
